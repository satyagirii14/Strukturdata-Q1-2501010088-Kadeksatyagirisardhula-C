```python
import sys
import platform

print("Python :", sys.version)
print("Platform:", platform.platform())

# BeautifulSoup di pip bernama beautifulsoup4
libraries = {
    "pandas": "pandas",
    "numpy": "numpy",
    "requests": "requests",
    "bs4": "beautifulsoup4"
}

for lib, package in libraries.items():
    try:
        m = __import__(lib)
        print(f"{lib:10s}: {getattr(m, '__version__', 'terpasang')}")
    except ImportError:
        print(f"{lib:10s}: BELUM TERPASANG -> Menjalankan instalasi...")
        get_ipython().system(f"pip install {package}")
```

```python
from pathlib import Path

ROOT = Path("proyek_akuisisi_data")
FOLDERS = [
    "data/raw",
    "data/interim",
    "data/processed",
    "notebooks",
    "src",
    "docs",
    "reports",
]

# Membuat folder secara rekursif
for f in FOLDERS:
    (ROOT / f).mkdir(parents=True, exist_ok=True)
    print("Dibuat:", ROOT / f)

# Membuat file .gitkeep agar folder kosong tetap terlacak oleh Git
for f in FOLDERS:
    (ROOT / f / ".gitkeep").touch()

print("\nStruktur folder proyek berhasil dibuat!")
```
